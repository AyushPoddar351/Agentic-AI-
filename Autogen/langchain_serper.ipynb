```python
import os
from dotenv import load_dotenv
load_dotenv()

os.getenv("SERPER_API_KEY")

from langchain_community.utilities import GoogleSerperAPIWrapper

search = GoogleSerperAPIWrapper()

search.run("Obama's first name?")
```

```python
from langchain_core.tools import Tool

# Create a single tool instead of a list of tools
serper_tool = Tool(
    name="Search",
    func=search.run,
    description="useful for when you need to answer questions about current events"
)
```

```python
from autogen_ext.tools.langchain import LangChainToolAdapter

# Create the LangChainToolAdapter with the single tool
search_tool = LangChainToolAdapter(serper_tool)

from autogen_agentchat.agents import AssistantAgent
from autogen_ext.models.ollama import OllamaChatCompletionClient
from autogen_agentchat.ui import Console

model_client = OllamaChatCompletionClient(model='llama3.1')

agent = AssistantAgent(
        "SearchAgent",
        tools=[search_tool],
        model_client=model_client,
        system_message="You are a web search agent."
)

# Example usage
# Uncomment to run
r = await Console(agent.run_stream(task="What is autogen?"))
print(r)
```
